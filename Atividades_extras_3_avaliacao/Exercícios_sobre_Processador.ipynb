#Questão 1
Descreva as funções básicas de uma UCP, indicando os seus componentes principais.

**Função de controle:** Associada às atividades de busca, interpretação e controle da execução das instruções, bem como o controle dos demais componentes do sistema. Principais componentes: Unidade de Controle (UC), Decodificador, Registrador de Instruções (RI), Contador de Instruções (CI), clock, Registrador de Endereço de Memória (REM) e Registrador de Dados de Memória (RDM). Vale ressaltar que o REM e o RDM podem ser recrutados tanto para a função de controle quanto para a de processamento, a depender da ação atual a ser executada pela CPU.

**Função processamento:** associada às atividades de efetiva execução de operações matemáticas, lógicas e outras. Manipula os dados indicados pela função de controle. Seus componentes principais são: a Unidade Aritmética e Lógica (UAL) e os registradores de uso geral.

# Questão 2
Quais são as funções da unidade aritmética e lógica - UAL?

Manipula os dados e executa as operações. Essas operações podem ser:

Aritméticas: soma, subtração, multiplicação, divisão.

Lógicas: AND, OR, XOR, NOT.

Outras: complementos, deslocamentos à esquerda e à direita.

# Questão 3
O que é e para que servia o ACC?

O ACC (que significa "acumulador") é um registrador de uso geral que faz a ligação da ULA com os demais dispositivos da CPU. Sua função principal consiste em armazenar dados temporários e os resultados intermediários das operações matemáticas e lógicas executadas pela ULA.

# Questão 5
Quais são as funções da unidade de controle de um processador?

Realizar a movimentação de dados de e para a CPU e gerenciar a ação da Unidade Aritmética e Lógica. Principais funções: busca de instruções, decodificação, emissão de sinais de controle, sincronização e gerenciamento do fluxo.

# Questão 6
Seria possível realizar o projeto de um processador em que o tamanho em bits do CI fosse diferente do tamanho do REM? Nesse caso, qual dos dois registradores deveria ter maior tamanho? Por quê?

Pelo modelo de von Neumann não seria possível, pois as instruções e os dados ocupam a mesma memória. Assim, ambos os registradores armazenam endereços e precisam ter o mesmo tamanho.

Contudo, em sistemas reais e mais avançados (como arquiteturas modernas com memória virtual ou paginação) onde o CI lida com endereços lógicos e o REM com endereços físicos seria possível.

# Questão 7
Considere um computador cuja MP é organizada com N células de 1 byte cada uma. As instruções interpretadas pela UCP possuem três tamanhos diferentes: as do tipo A possuem 16 bits; as do tipo B têm 32 bits e as do tipo C possuem 48 bits. Considerando que o código de operação de cada uma tem um tamanho fixo e igual a 8 bits e que os programas executados nesse processador são constituídos de uma mistura dos três tipos de instruções, imagine um processo prático para incremento automático do CI após a execução de cada instrução de um programa.

C.Op. = 8 bits -> 2^8 = 256 instruções

Como as instruções possuem tamanhos variáveis, o incremento do CI não será fixo. O incremento irá variar conforme o tipo de instrução a ser executada. Assim que a CPU buscar o primeiro byte na memória, irá identificar o tipo de instrução e incrementar o CI conforme esta.

É necessário converter o tamanho das instruções de bits para bytes (tamanho da célula) para saber quanto o CI deve "saltar" ao incrementar:

16 bits/8 -> 2B: CI <- CI + 2

32 bits/8 -> 4B: CI <- CI + 4

48 bits/8 -> 6B: CI <- CI + 6

## Item A
a) Qual deverá ser a largura, em bits, do campo C.Op. das instruções?

Inicialmente, o RI armazena a instrução em execução 2C4F08D9.

Cada dígito hexadecimal equivale a 4 bits em binário. Portanto, o tamanho total da instrução é: 8 x 4 = **32 bits**.

**Largura do OP.2:** Sabendo que todos os endereços possuem 16 bits e o campo OP.2 é utilizado para armazenar endereços, ele precisa ter exatamente 16 bits de largura para comportar um endereço.

**Largura do REG:** Conforme o enunciado, o processador possui um conjunto de 16 registradores de dados. Para identificar cada um desses 16 registradores, a quantia de bits necessária é dada pelo logaritmo na base 2 da quantidade de Registradores -> Log2(16) = 4.

Portanto, a largura em bits, do campo C.OP. é determinada por:

32 bits (total de bits da instrução) - 16 bits (OP.2) - 4 bits (REG) = 12 bits.

Logo, a Largura em bits do campo C.OP. é 12 bits.

## Item B
b) Qual deverá ser a largura do campo Reg das instruções?

Conforme o enunciado, o processador possui um conjunto de 16 registradores de dados. Então, para identificar cada um desses 16 registradores, a quantia necessária é de 4 bits, uma vez que:

2^4 = 16

## Item C
Quantos acessos à memória devem ser realizados para executar o ciclo completo da instrução de C.Op. igual a 11? E para a instrução de C.Op. igual a 21?

**Para a instrução C.OP. = 11(JP R, OP.):**

Se R > 0, então CI <- OP. (Desvio condicional).

Na busca, será feito 1 acesso para ler a instrução da memória e trazê-la para o RI.

Na execução, a CPU faz uma verificação interna no registrador R e apenas atualiza o valor do Contador de instruções (CI). Nenhuma leitura ou escrita de dados externos são realizados na memória.


Total de acessos: 1 acesso.


**Para a instrução C.OP. = 21(SUB R, OP.):**

(R) <- (R) - OP. (Subtração).

Na busca, ocorre 1 acesso para ler a instrução da memória.

A instrução SUB não usa parênteses no operando, o que indica que se trata de um operando imediato, do próprio valor já está embutido no campo OP. 2 da instrução. Sendo assim, a CPU não precisa fazer um acesso adicional para buscar os dados.

Total de acessos: 1 acesso.
